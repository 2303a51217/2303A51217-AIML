```python
import matplotlib.pyplot as plt

x1 = -1
x2 = 1
import numpy as np
from scipy import stats
from cProfile import label


x = np.linspace(-4, 4, 1000)
y = stats.norm.pdf(x)
plt.plot(x, y, label='normal distribution')
plt.fill_between(x, y, where=(x > x1) & (x < x2), color='red', alpha=0.5 , label='P(x1 < x < x2)')

plt.fill_between(x, y, where=(x < x2), color='blue', alpha=0.3 , label='P(x < x2)')

plt.fill_between(x, y, where=(x > x1), color='green', alpha=0.3 , label='P(x1 > x)')

plt.xlabel('x')
plt.ylabel('probability density')
plt.title('Normal Distribution with shaded area')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import  numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
mu = 100
sigma = 5
data = np.random.normal(mu, sigma, 10000)
x = np.linspace(mu - 3*sigma, mu + 3*sigma, 10000)
pdf = norm.pdf(x, mu, sigma)

plt.figure(figsize=(10, 6))
plt.hist(data, bins=30, density=True, alpha=0.6, color='#1f7', edgecolor='black', linestyle='dashed', linewidth=1.5)
plt.plot(x, pdf, 'r-', lw=2)
plt.title('Normal Distribution: $\mu$=100, $\sigma')
plt.xlabel('Value')
plt.ylabel('Density')
plt.show()

```
